##**Conhecendo os dados**

### Importando os dados

```python
import pandas as pd
```

```python
df = pd.read_csv('imigrantes_canada.csv')
df.info()
```

### Realizando alguns tratamentos

```python
df.set_index('País', inplace=True)
anos = list(map(str, range(1980, 2014)))
anos
```

### Extraindo a série de dados para o Brasil

```python
brasil = df.loc['Brasil', anos]
brasil_dict = {'ano': brasil.index.tolist(), 'imigrantes': brasil.values.tolist()}

dados_brasil = pd.DataFrame(brasil_dict)
dados_brasil
```

```python
import matplotlib.pyplot as plt
```

## Evolução de imigrantes brasileiros para o Canadá

```python
fig, ax = plt.subplots()
ax.plot(dados_brasil['ano'], dados_brasil['imigrantes'], lw=3)
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_title('Imigração do Brasil para o Canadá\n1980 a 2013', fontsize=18, loc='left')
ax.set_xlabel('Ano', fontsize=14)
ax.set_ylabel('Número de imigrantes', fontsize=14)
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)


plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Dados de exemplo (substitua pelo seu DataFrame real da Alura)
dados = {
    "País": [
        "Guiana",
        "Colômbia",
        "Peru",
        "Brasil",
        "Chile",
        "Venezuela",
        "Argentina",
        "Equador",
        "Uruguai",
        "Bolívia",
        "Paraguai",
        "Suriname",
    ],
    "Imigrantes": [
        75785,
        72088,
        32652,
        29659,
        21359,
        21267,
        19596,
        12233,
        4715,
        3205,
        1944,
        739,
    ],
}

df_sul_america = pd.DataFrame(dados)

# Inverte a ordem para que o maior país fique no topo do gráfico de barras horizontais
df_sul_america = df_sul_america.iloc[::-1]

# 2. Configuração da figura
fig, ax = plt.subplots(figsize=(10, 6))

# 3. Definição de cores (Destacando o Brasil em verde e os outros em cinza claro)
cores = [
    "green" if pais == "Brasil" else "lightgray"
    for pais in df_sul_america["País"]
]

# 4. Construção do gráfico de barras horizontais
barras = ax.barh(
    df_sul_america["País"], df_sul_america["Imigrantes"], color=cores
)

# 5. Adicionando os rótulos de dados (valores numéricos) ao lado de cada barra
ax.bar_label(barras, padding=3, fontsize=9)

# 6. Título formatado e alinhado à esquerda
ax.set_title(
    "América do Sul: Brasil foi o quarto país com mais imigrantes\n"
    "para o Canadá no período de 1980 a 2013",
    fontsize=14,
    loc="left",
    pad=20,
)

# 7. Limpeza visual: remove as bordas (spines) e o eixo X inferior
for spine in ["top", "right", "bottom", "left"]:
    ax.spines[spine].set_visible(False)

ax.xaxis.set_visible(False)  # Oculta a linha do eixo X
ax.tick_params(axis="y", left=False)  # Remove os tracinhos do eixo Y

# 8. Ajuste fino de layout
plt.tight_layout()

# Exibe o gráfico
plt.show()
```
